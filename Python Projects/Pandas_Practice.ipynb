```python
#To load dataframe it sould be 2d so we should read as dictionary key value paris 
students_record = {"Name": ['ABC', 'DEF', 'GHI', 'JKL', 'MNO','pqr','stu'],
         "Marks":[100, 95, 80, 90, 100, 90, 80],
         "Age":[20, 21, 22, 23, 20, 22, 23]}
print(students_record)
print(type(students_record))
```

```python
df = pd.DataFrame(students_record)
print(df)
```

```python
#Common Operations -- Pandas Basic operations on Table:

print("====================Head OP====================")
print(df.head(3))  #Head by default display first 5 rows:
print("====================Tail OP====================")
print(df.tail())
print("====================df.columns====================")
print(df.columns)
print("====================df.dtypes====================")
print(df.dtypes)
print("====================MIN OP====================")
print(df.min())
print("====================MAX OP====================")
print(df.max())
print("====================INFO====================")
print(df.info())
print("====================DESCRIBE====================")
print(df.describe())
print("====================Accessing columns====================")
print(df['Marks'])
print("====================Filter rows (Marks greater than 80 - T/F)====================")
print(df['Marks']>80)
print("====================Filter rows (Marks greater than 80 - display only filtered rows)====================")
print(df[df['Marks']>80])
print(f"**********Sum of marks: {df['Marks'].sum()}")
print("====================df display====================")
print(df)
print("====================df.shape****R:(rows,columns)====================")
print(df.shape)
print("====================pick 1 column====================")
print(df["Name"])
print("====================pick multiple columns====================")
print(df[['Name','Age']])
print("====================sorting default = Asc====================")
print(df.sort_values('Age'))
print("====================sorting desc====================")
print(df.sort_values('Marks',ascending=False))
print("====================Adding new column====================")
print(f"df shape before adding new column: {df.shape}")
df['New_Column'] = ['1st','2nd','3rd','4th','5th','','7th']
print(f"df shape after adding new column: {df.shape}")
print(df)
print("====================Export the data to CSv file with/without Index====================")
df.to_csv('Vanitha_practise_dataset.csv')
df.to_csv('Vanitha_practise_dataset2.csv',index=False)
print("************END****************************")
```
